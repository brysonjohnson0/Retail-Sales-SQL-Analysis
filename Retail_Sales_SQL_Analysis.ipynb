# SQL Analysis

##**Business Question 1**

Which products generated the highest total sales revenue?

This analysis identifies the company's top-performing products based on total revenue.

```python
query= """
SELECT
   "Product Name" AS Product,
   ROUND(SUM(Sales),2) AS Total_Sales
FROM sales
GROUP BY "Product Name"
ORDER BY Total_Sales DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

###Findings
Canon imageCLASS 2200 Advanced Copier was the top-selling product.
The top 10 products contributed a significant portion of total revenue.

###Business Insight

High-performing products should be monitored closely for inventory planning, pricing strategy, and promotional opportunities.

## **Business Question 2**

Which region generated the highest total sales?

Understanding regional performance helps identify where the business generates the most revenue.

```python
query= """
SELECT
   Region,
   ROUND(SUM(Sales),2) AS Total_Sales
FROM Sales
GROUP BY Region
ORDER BY Total_Sales DESC
LIMIT 5;
"""

pd.read_sql(query, conn)
```

##Findings
The West region generated the highest sales.
The South generated the lowest sales.

##Business Insight

The West represents the company's strongest market. Lower-performing regions could benefit from targeted marketing campaigns or operational improvements.

## **Busines Question 3**

Which customers generated the highest total sales?

This analysis compares sales performance across the company's top customer's.

```python
query= """
SELECT
   "Customer ID",
   ROUND(SUM(Sales),2) AS Total_Sales
FROM Sales
GROUP BY "Customer ID"
ORDER BY Total_Sales DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

##Findings
A relatively small group of customers generated the highest sales.

##Business Insight

These customers may represent strategic accounts. Customer retention efforts and personalized engagement could help maintain long-term revenue.

## **Business Question 4**

Which product categories generated the most sales?

This analysis compares sales performance across the company's primary product categories.

```python
query= """
SELECT
   Category,
   ROUND(SUM(Sales),2) AS Total_Sales
FROM Sales
GROUP BY Category
ORDER BY Total_Sales DESC;
"""

pd.read_sql(query, conn)
```

##Findings
Technology generated the highest sales.
Furniture ranked second.
Office Supplies ranked third.

##Business Insight

Technology appears to be the company's strongest revenue category. Future analysis could compare sales with profitability to determine whether high sales also translate into strong margins.

## **Business Question 5**

What does the company's monthly trend in revenue look like?

This analysis compares sales performance month-to-month.

```python
query= """
SELECT
   strftime('%Y-%m', "Order Date") AS Month,
   ROUND(SUM(Sales),2) AS Total_Sales
FROM Sales
GROUP BY Month
ORDER BY Month;
"""

pd.read_sql(query, conn)
```

##Findings
Sales fluctuated throughout the year.
Overall revenue increased over the four-year period.

##Business Insight

The company experienced long-term growth while also showing seasonal fluctuations. Understanding these trends can support forecasting and inventory planning.

## **Business Question 6**

Which year did the company generate the most revenue?

This analysis compares sales performance year-to-year.

```python
query= """
SELECT
   strftime('%Y', "Order Date") AS Year,
   ROUND(SUM(Sales),2) AS Total_Sales
FROM Sales
GROUP BY Year
ORDER BY Year;
"""

pd.read_sql(query, conn)
```

##Findings
2017 generated the highest annual sales.
Sales generally increased from 2014 through 2017.

##Business Insight

The steady upward trend suggests sustained business growth. Additional data would be needed to determine what factors contributed to the growth.

## **Business Question 7**

Which customer segment generated the most revenue?

This analysis compares which customer groups contributed the most revenue.

```python
query= """
SELECT
   Segment,
   ROUND(SUM(Sales),2) AS Total_Sales
FROM Sales
GROUP BY Segment
ORDER BY Segment;
"""

pd.read_sql(query, conn)
```

##Findings
Consumer was the highest-revenue segment.
Home Office generated the lowest revenue.

##Business Insight

The Consumer segment should remain a key focus for marketing and customer retention. At the same time, the Home Office segment may present opportunities for targeted growth initiatives.

## **Business Question 8**

Which states generated the highest sales revenue?

This analysis compares which states contributed the most revenue.

```python
query= """
SELECT
   State,
   ROUND(SUM(Sales),2) AS Total_Sales
FROM Sales
GROUP BY State
ORDER BY Total_Sales DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

##Findings
California generated the highest sales.
New York ranked second.
Texas ranked third.

##Business Insight

These states represent major revenue markets and should remain priorities for sales strategy, customer engagement, and resource allocation.

## **Business Question 9**

Which states generated the highest total profit?

This analysis compares which states contributed the most profit.

```python
query= """
SELECT
   State,
   ROUND(SUM(Profit),2) AS Profit
FROM Sales
GROUP BY State
ORDER BY Profit DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

##Findings
California generated the highest total profit.
New York ranked second.
Washington ranked third.

##Business Insight

Comparing sales and profit by state helps identify markets that not only generate high revenue but also contribute strong profitability. This distinction is valuable when making investment and expansion decisions.
